```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
```

```python
url = "https://raw.githubusercontent.com/shivang98/Social-Network-ads-Boost/master/Social_Network_Ads.csv"

# Load the data into a pandas DataFrame
df = pd.read_csv(url)

# Display the first 5 rows to make sure it loaded correctly
df.head()
```

```python
def analyzer(max_depth):
    data = df
    X = data.iloc[:, [2, 3]].values
    y = data.iloc[:, -1].values

    clf = DecisionTreeClassifier(max_depth=max_depth)
    clf.fit(X, y)

    a = np.arange(start=X[:, 0].min() - 1, stop=X[:, 0].max() + 1, step=0.1)
    b = np.arange(start=X[:, 1].min() - 1, stop=X[:, 1].max() + 1, step=100)

    XX, YY = np.meshgrid(a, b)

    input_array = np.array([XX.ravel(), YY.ravel()]).T
    labels = clf.predict(input_array)

    plt.contourf(XX, YY, labels.reshape(XX.shape), alpha=0.5)
    plt.scatter(X[:, 0], X[:, 1], c=y)
    plt.show()
```

```python
analyzer(max_depth=None)


```

```python
analyzer(max_depth=1)

```

```python
analyzer(max_depth=4)
```
