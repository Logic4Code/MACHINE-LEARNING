```python
import numpy as np
import pandas as pd
```

```python
data_url = "http://lib.stat.cmu.edu/datasets/boston"

# 2. Read the raw text file (skip the description rows)
boston_raw = pd.read_csv(data_url, sep=r"\s+", skiprows=22, header=None)

# 3. Format the data (the original file splits 13 features across alternating lines)
X = np.hstack([boston_raw.values[::2, :], boston_raw.values[1::2, :2]])
y = boston_raw.values[1::2, 2]

# 4. Define the official Boston dataset feature names manually
feature_names = ['CRIM', 'ZN', 'INDUS', 'CHAS', 'NOX', 'RM', 'AGE', 'DIS', 'RAD', 'TAX', 'PTRATIO', 'B', 'LSTAT']

# 5. Create the unified pandas DataFrame
df = pd.DataFrame(X, columns=feature_names)
df['MEDV'] = y  # Add the target variable (Median value of owner-occupied homes)

# 6. View the first 5 rows
df.head()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.model_selection import cross_val_score
     

lr = LinearRegression()
dt = DecisionTreeRegressor()
svr = SVR()
     

estimators = [('lr',lr),('dt',dt),('svr',svr)]
     

for estimator in estimators:
  scores = cross_val_score(estimator[1],X,y,scoring='r2',cv=10)
  print(estimator[0],np.round(np.mean(scores),2))
     
```

```python

from sklearn.ensemble import VotingRegressor
     

vr = VotingRegressor(estimators)
scores = cross_val_score(vr,X,y,scoring='r2',cv=10)
print("Voting Regressor",np.round(np.mean(scores),2))
```

```python

for i in range(1,4):
  for j in range(1,4):
    for k in range(1,4):
      vr = VotingRegressor(estimators,weights=[i,j,k])
      scores = cross_val_score(vr,X,y,scoring='r2',cv=10)
      print("For i={},j={},k={}".format(i,j,k),np.round(np.mean(scores),2))
```

```python

# using the same algorithm

dt1 = DecisionTreeRegressor(max_depth=1)
dt2 = DecisionTreeRegressor(max_depth=3)
dt3 = DecisionTreeRegressor(max_depth=5)
dt4 = DecisionTreeRegressor(max_depth=7)
dt5 = DecisionTreeRegressor(max_depth=None)
     

estimators = [('dt1',dt1),('dt2',dt2),('dt3',dt3),('dt4',dt4),('dt5',dt5)]
     

for estimator in estimators:
  scores = cross_val_score(estimator[1],X,y,scoring='r2',cv=10)
  print(estimator[0],np.round(np.mean(scores),2))
     
```

```python

vr = VotingRegressor(estimators)
scores = cross_val_score(vr,X,y,scoring='r2',cv=10)
print("Voting Regressor",np.round(np.mean(scores),2))
```
