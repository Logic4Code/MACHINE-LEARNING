```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
```

```python
url = "https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/main/day40-iterative-imputer/50_Startups.csv"

# Importing the CSV file
df = np.round(pd.read_csv(url)[['R&D Spend','Administration','Marketing Spend','Profit']]/10000)
```

```python
df = df.sample(5)
df
```

```python
df = df.iloc[:,0:-1]
df
```

```python
df.iloc[1,0] = np.nan
df.iloc[3,1] = np.nan
df.iloc[-1,-1] = np.nan
df.head()
```

```python
# Step 1 - Impute all missing values with mean of respective col

df0 = pd.DataFrame()

df0['R&D Spend'] = df['R&D Spend'].fillna(df['R&D Spend'].mean())
df0['Administration'] = df['Administration'].fillna(df['Administration'].mean())
df0['Marketing Spend'] = df['Marketing Spend'].fillna(df['Marketing Spend'].mean())
# 0th Iteration
df0
```

```python
# Remove the col1 imputed value
df1 = df0.copy()

df1.iloc[1,0] = np.nan

df1
```

```python
# Use first 3 rows to build a model and use the last for prediction

X = df1.iloc[[0,2,3,4],1:3]
X
```

```python
y = df1.iloc[[0,2,3,4],0]
y
```

```python

lr = LinearRegression()
lr.fit(X,y)
lr.predict(df1.iloc[1,1:].values.reshape(1,2))
```

```python
df1.iloc[1,0] = 6.10888575
df1
```

```python
# Remove the col2 imputed value

df1.iloc[3,1] = np.nan

df1
```

```python
# Use last 3 rows to build a model and use the first for prediction
X = df1.iloc[[0,1,2,4],[0,2]]
X
```

```python
y = df1.iloc[[0,1,2,4],1]
y
```

```python
lr = LinearRegression()
lr.fit(X,y)
lr.predict(df1.iloc[3,[0,2]].values.reshape(1,2))
```

```python
df1.iloc[3,1] = 19.06475122
df1
```

```python
# Remove the col3 imputed value
df1.iloc[4,-1] = np.nan

df1
```

```python
# Use last 3 rows to build a model and use the first for prediction
X = df1.iloc[0:4,0:2]
X
```

```python
y = df1.iloc[0:4,-1]
y
```

```python
lr = LinearRegression()
lr.fit(X,y)
lr.predict(df1.iloc[4,0:2].values.reshape(1,2))
```

```python
df1.iloc[4,-1] = 11.86436955
# After 1st Iteration
df1
```

```python
# Subtract 0th iteration from 1st iteration

df1 - df0
```

```python
df2 = df1.copy()

df2.iloc[1,0] = np.nan

df2
```

```python
X = df2.iloc[[0,2,3,4],1:3]
y = df2.iloc[[0,2,3,4],0]

lr = LinearRegression()
lr.fit(X,y)
lr.predict(df2.iloc[1,1:].values.reshape(1,2))
```

```python
df2.iloc[1,0] = 4.17335709
df2.iloc[3,1] = np.nan
X = df2.iloc[[0,1,2,4],[0,2]]
y = df2.iloc[[0,1,2,4],1]

lr = LinearRegression()
lr.fit(X,y)
lr.predict(df2.iloc[3,[0,2]].values.reshape(1,2))
```

```python
df2.iloc[3,1] = 21.3713218
df2.iloc[4,-1] = np.nan

X = df2.iloc[0:4,0:2]
y = df2.iloc[0:4,-1]

lr = LinearRegression()
lr.fit(X,y)
lr.predict(df2.iloc[4,0:2].values.reshape(1,2))
```

```python
df2.iloc[4,-1] = 11.87540753
df2
```

REPEAT TILL DIFFERENCE IS 0 OR NEAR 0


```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
from sklearn.linear_model import LinearRegression

mice_imputer = IterativeImputer(estimator=LinearRegression(), max_iter=10, random_state=42)

imputed_array = mice_imputer.fit_transform(df)
df_imputed = pd.DataFrame(imputed_array, columns=df.columns)

print("\nMissing values after MICE imputation:")
print(df_imputed.isnull().sum())
```
